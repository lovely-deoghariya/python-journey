```python
#write a nested for loop to print a 3×3 grid of *
for i in range(3):
    for j in range(3):
        print("*", end=" ")
    print()
```

```python
#write a nested loop to print numbers in a 4×4 square
for i in range(1, 5):
    for j in range(1, 5):
        print(j, end=" ")
    print()
```

```python
#create a pattern of five rows using #
for i in range(5):
    for j in range(5):
        print("#", end=" ")
    print()
```

```python
#print a multiplication table from 1 to 5 using nested loops
for i in range(1, 6):
    for j in range(1, 11):
        print(i, "x", j, "=", i*j)
    print()
```

```python
#Display all coordinate pairs (rows, column) from 1 to 3
for row in range(1, 4):
    for column in range(1, 4):
        print("(", row, ",", column, ")")
```

```python
#print a right angle triangle using nested loops
for i in range(1, 6):
    for j in range(i):
        print("*", end=" ")
    print()
```

```python
#count how many times the inner loop runs when the outer loop runs for times
count = 0

for i in range(4):
    for j in range(3):
        count += 1

print(count)
```

```python
#create a rectangle of @ symbols with 4 rows and 6 columns
for i in range(4):
    for j in range(6):
        print("@", end=" ")
    print()
```

```python
#print only the rows numbers before each rows of stars
for i in range(1, 6):
    print(i, end=" ")
    for j in range(5):
        print("*", end=" ")
    print()
```

```python
#Revision: write a for loop using range() to print numbers from 10 down to 1
for i in range(10, 0, -1):
    print(i)
```
