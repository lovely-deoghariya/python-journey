```python
#Today's new concept: mini list project 2
```

```python
#create fruit list
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
```

```python
#add one fruit
fruits = ["Apple", "Banana", "Mango"]
fruits.append("Orange")
print(fruits)
```

```python
#sort the list
fruits = ["Mango", "Apple", "Banana"]
fruits.sort()
print(fruits)
```

```python
#Reverse the list
fruits = ["Apple", "Banana", "Mango"]
fruits.reverse()
print(fruits)
```

```python
#count "Apple"
fruits = ["Apple", "Banana", "Apple", "Mango"]
print(fruits.count("Apple"))
```

```python
#find index of "Banana"
fruits = ["Apple", "Banana", "Mango"]
print(fruits.index("Banana"))
```

```python
#copy the list
fruits = ["Apple", "Banana", "Mango"]
new_list = fruits.copy()
print(new_list)
```

```python
#join items into one string
fruits = ["Apple", "Banana", "Mango"]
result = ", ".join(fruits)
print(result)
```

```python
#print list length
fruits = ["Apple", "Banana", "Mango"]
print(len(fruits))
```
