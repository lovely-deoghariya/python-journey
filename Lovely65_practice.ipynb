```python
#Today's new concept: mini list challenge
```

```python
#create fruit list
fruits = ["Apple", "Banana", "Mango"]
print(fruits)
```

```python
#Add one fruit
fruits = ["Apple", "Banana", "Mango"]
fruits.append("Orange")
print(fruits)
```

```python
#insert one fruit
fruits = ["Apple", "Banana", "Mango"]
fruits.insert(1, "Grapes")
print(fruits)
```

```python
#remove one fruit
fruits = ["Apple", "Banana", "Mango"]
fruits.remove("Banana")
print(fruits)
```

```python
#sort the list
fruits = ["Mango", "Apple", "Banana"]
fruits.sort()
print(fruits)
```

```python
#reverse the list
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

```python
#print each items using
fruits = ["Apple", "Banana", "Mango"]
print(len(fruits))
```

```python
#check "Apple" in list
fruits = ["Apple", "Banana", "Mango"]
print("Apple" in fruits)
```
