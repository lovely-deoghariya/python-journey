```python
#Today's new concept: copy a list
```

```python
#create fruit list
fruits = ["Apple", "Mango", "Banana"]
print(fruits)
```

```python
#copy using copy()
fruits = ["Apple", "Mango", "Banana"]
copy_fruits = fruits.copy()
print(copy_fruits)
```

```python
#print both lists
fruits = ["Apple", "Mango", "Banana"]
copy_fruits = fruits.copy()

print(fruits)
print(copy_fruits)
```

```python
#Add one fruit to copied list
fruits = ["Apple", "Mango", "Banana"]
copy_fruits = fruits.copy()

copy_fruits.append("Orange")
print(copy_fruits)
```

```python
#print copied list
fruits = ["Apple", "Mango", "Banana"]
copy_fruits = fruits.copy()

print(copy_fruits)
```
